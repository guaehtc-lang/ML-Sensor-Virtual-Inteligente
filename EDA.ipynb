```python
import matplotlib.pyplot as plt
import pandas as pd

df = nivel_vb01.copy()

df["Valor proceso LT411 Time"] = pd.to_datetime(df["Valor proceso LT411 Time"], dayfirst=True)

plt.figure(figsize=(12, 4))
plt.plot(
    df["Valor proceso LT411 Time"],
    df["Valor proceso LT411 ValueY"],
    linewidth=1.2
)
plt.title("Oscilación del nivel LT-411")
plt.xlabel("Tiempo")
plt.ylabel("Nivel (%)")
plt.tight_layout()
plt.show()
```

```python

df["Valor proceso LT411 Time"] = pd.to_datetime(
    df["Valor proceso LT411 Time"],
    dayfirst=True
)

inicio = df["Valor proceso LT411 Time"].min()
fin = inicio + pd.Timedelta(hours=1)

df_1h = df[
    (df["Valor proceso LT411 Time"] >= inicio) &
    (df["Valor proceso LT411 Time"] <= fin)
]

plt.figure(figsize=(12, 4))
plt.plot(
    df_1h["Valor proceso LT411 Time"],
    df_1h["Valor proceso LT411 ValueY"],
    linewidth=1.2
)
plt.title("Nivel LT-411 — primera hora")
plt.xlabel("Tiempo")
plt.ylabel("Nivel (%)")
plt.tight_layout()
plt.show()
```

```python
datasets = {
    "nivel_vb01": nivel_vb01,
    "den_int_vb01": den_int_vb01,
    "vapor_he01": vapor_he01,
    "vacio": vacio,
    "temp_salmuera_he01": temp_salmuera_he01,
    "nivel_caudal_pct02": nivel_caudal_pct02,
    "presion_vb01_he01": presion_vb01_he01,
    "tt_413": tt_413
}
```

```python
resumen_numericas = []

for nombre_df, df in datasets.items():
    
    columnas_numericas = df.select_dtypes(include="number").columns
    
    for columna in columnas_numericas:
        resumen_numericas.append({
            "dataframe": nombre_df,
            "columna": columna,
            "min": df[columna].min(),
            "max": df[columna].max(),
            "media": df[columna].mean(),
            "std": df[columna].std()
        })

resumen_numericas = pd.DataFrame(resumen_numericas)

resumen_numericas
```

```python
datasets = [
    ("nivel_vb01", nivel_vb01),
    ("den_int_vb01", den_int_vb01),
    ("vapor_he01", vapor_he01),
    ("vacio", vacio),
    ("temp_salmuera_he01", temp_salmuera_he01),
    ("nivel_caudal_pct02", nivel_caudal_pct02),
    ("presion_vb01_he01", presion_vb01_he01),
    ("tt_413", tt_413)
]
```

```python
resumen_fechas = []

for nombre, df in datasets:
    
    columnas_time = [col for col in df.columns if "Time" in col]
    
    # Usamos la primera columna Time como referencia
    col_time = columnas_time[0]
    
    # Convertimos a datetime
    df[col_time] = pd.to_datetime(df[col_time], dayfirst=True, errors="coerce")
    
    resumen_fechas.append({
        "dataframe": nombre,
        "columna_time": col_time,
        "filas": df.shape[0],
        "inicio": df[col_time].min(),
        "fin": df[col_time].max(),
        "duplicados_time": df[col_time].duplicated().sum()
    })

resumen_fechas = pd.DataFrame(resumen_fechas)

resumen_fechas
```

```python
resumen_frecuencia = []

for nombre, df in datasets:
    
    columnas_time = [col for col in df.columns if "Time" in col]
    col_time = columnas_time[0]
    
    df[col_time] = pd.to_datetime(df[col_time], dayfirst=True, errors="coerce")
    
    diferencias = df[col_time].sort_values().diff()
    
    resumen_frecuencia.append({
        "dataframe": nombre,
        "columna_time": col_time,
        "frecuencia_mas_comun": diferencias.mode()[0],
        "min_diferencia": diferencias.min(),
        "max_diferencia": diferencias.max(),
        "saltos_distintos_a_1s": (diferencias.dropna() != pd.Timedelta(seconds=1)).sum()
    })

resumen_frecuencia = pd.DataFrame(resumen_frecuencia)

resumen_frecuencia
```

```python
resumen_time_interno = []

for nombre, df in datasets:
    
    columnas_time = [col for col in df.columns if "Time" in col]
    
    # Convertir TODAS las columnas Time a datetime
    for col in columnas_time:
        df[col] = pd.to_datetime(df[col], dayfirst=True, errors="coerce")
    
    df_time = df[columnas_time]
    
    son_iguales = df_time.eq(df_time.iloc[:, 0], axis=0).all().all()
    
    resumen_time_interno.append({
        "dataframe": nombre,
        "columnas_time": len(columnas_time),
        "time_interno_igual": son_iguales
    })

resumen_time_interno = pd.DataFrame(resumen_time_interno)

resumen_time_interno
```
