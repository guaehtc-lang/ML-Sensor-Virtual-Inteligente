```python
import pandas as pd
```

```python
nivel_vb01 = pd.read_csv('data/Nivel_VB-01.csv', encoding='utf-16', sep=';', decimal=',')
den_int_vb01 = pd.read_csv('data/Den-Int_VB-01.csv', encoding='utf-16', sep=';', decimal=',')
vapor_he01 = pd.read_csv('data/Vapor_HE-01.csv', encoding='utf-16', sep=';', decimal=',')
vacio = pd.read_csv('data/Vacío.csv', encoding='utf-16', sep=';', decimal=',')
temp_salmuera_he01 = pd.read_csv('data/Temperatura_salmuera_HE01.csv', encoding='utf-16', sep=';', decimal=',')
nivel_caudal_pct02 = pd.read_csv('data/Nivel_Caudal_PCT-02.csv', encoding='utf-16', sep=';', decimal=',')
presion_vb01_he01 = pd.read_csv('data/Presion_VB01_HE01.csv', encoding='utf-16', sep=';', decimal=',')
```

```python
nivel_limpio = nivel_vb01[[
    "Valor proceso LT411 Time",
    "Valor proceso LT411 ValueY",
    "Set point ValueY",
    "Accion (valvula) LV411 ValueY"
]].copy()

nivel_limpio.columns = [
    "Time",
    "LT411",
    "SP_LT411",
    "LV411"
]

nivel_limpio["Time"] = pd.to_datetime(nivel_limpio["Time"], dayfirst=True)

nivel_limpio.head()
```

```python
den_int_limpio = den_int_vb01[[
    "Valor densidad DT412 Time",
    "Valor densidad DT412 ValueY",
    "Valor intensidad P-101 ValueY"
]].copy()

den_int_limpio.columns = [
    "Time",
    "DT412",
    "INT_P101"
]

den_int_limpio["Time"] = pd.to_datetime(den_int_limpio["Time"], dayfirst=True)

den_int_limpio.head()
```

```python
vapor_limpio = vapor_he01[[
    "Caudal vapor FQC-400-1 Time",
    "Caudal vapor FQC-400-1 ValueY",
    "Set point ValueY",
    "Acción válvula FV400-1 ValueY"
]].copy()

vapor_limpio.columns = [
    "Time",
    "FQC400_1",
    "SP_VAPOR",
    "FV400_1"
]

vapor_limpio["Time"] = pd.to_datetime(vapor_limpio["Time"], dayfirst=True)

vapor_limpio.head()
```

```python
vacio_limpio = vacio[[
    "Valor proceso PT442 Time",
    "Valor proceso PT442 ValueY",
    "Set point ValueY",
    "Accion (valvula) PV442 ValueY"
]].copy()

vacio_limpio.columns = [
    "Time",
    "PT442",
    "SP_PT442",
    "PV442"
]

vacio_limpio["Time"] = pd.to_datetime(vacio_limpio["Time"], dayfirst=True)

vacio_limpio.head()
```

```python
temp_limpio = temp_salmuera_he01[[
    "Temp TT413 Time",
    "Temp TT413 ValueY",
    "Temp TT415 ValueY"
]].copy()

temp_limpio.columns = [
    "Time",
    "TT413",
    "TT415"
]

temp_limpio["Time"] = pd.to_datetime(temp_limpio["Time"], dayfirst=True)

temp_limpio.head()
```

```python
pct02_limpio = nivel_caudal_pct02[[
    "Valor proceso LT426 Time",
    "Valor proceso LT426 ValueY",
    "Set point ValueY",
    "Accion (valvula) ValueY",
    "Caudal FT428 ValueY"
]].copy()

pct02_limpio.columns = [
    "Time",
    "LT426",
    "SP_LT426",
    "ACCION_PCT02",
    "FT428"
]

pct02_limpio["Time"] = pd.to_datetime(pct02_limpio["Time"], dayfirst=True)

pct02_limpio.head()
```

```python
presion_limpio = presion_vb01_he01[[
    "Valor proceso PIT-410 Time",
    "Valor proceso PIT-410 ValueY",
    "Valor proceso PIT-414 ValueY"
]].copy()

presion_limpio.columns = [
    "Time",
    "PIT410",
    "PIT414"
]

presion_limpio["Time"] = pd.to_datetime(presion_limpio["Time"], dayfirst=True)

presion_limpio.head()
```

```python
dataset_limpio = nivel_limpio.merge(den_int_limpio, on="Time", how="inner")

dataset_limpio = dataset_limpio.merge(vapor_limpio, on="Time", how="inner")
dataset_limpio = dataset_limpio.merge(vacio_limpio, on="Time", how="inner")
dataset_limpio = dataset_limpio.merge(temp_limpio, on="Time", how="inner")
dataset_limpio = dataset_limpio.merge(pct02_limpio, on="Time", how="inner")
dataset_limpio = dataset_limpio.merge(presion_limpio, on="Time", how="inner")

dataset_limpio.head()
```

```python
dataset_limpio = dataset_limpio.sort_values("Time")

dataset_limpio.head()
```

```python
dataset_limpio = dataset_limpio.reset_index(drop=True)

dataset_limpio.head()
```

```python
dataset_limpio.to_csv("data_limpio/dataset_unificado_1s.csv", index=False)
```

```python
dataset_1s = pd.read_csv("data_limpio/dataset_unificado_1s.csv")

dataset_1s["Time"] = pd.to_datetime(dataset_1s["Time"])

```

```python
print("Filas y columnas:", dataset_1s.shape)
print("Inicio:", dataset_1s["Time"].min())
print("Fin:", dataset_1s["Time"].max())
print("Duplicados Time:", dataset_1s["Time"].duplicated().sum())
print("Nulos totales:", dataset_1s.isnull().sum().sum())
```

```python
df = dataset_1s.copy()

df = df.set_index("Time")

lt411_5s = df["LT411"].resample("5s").mean()
lt411_10s = df["LT411"].resample("10s").mean()
lt411_30s = df["LT411"].resample("30s").mean()
lt411_1min = df["LT411"].resample("1min").mean()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 4))
plt.plot(df.index, df["LT411"], linewidth=0.5, label="1s")
plt.plot(lt411_1min.index, lt411_1min, linewidth=1.2, label="1min")

plt.title("LT-411: señal original 1s vs remuestreo 1min")
plt.xlabel("Tiempo")
plt.ylabel("Nivel (%)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
inicio = df.index.min()
fin = inicio + pd.Timedelta(hours=1)

plt.figure(figsize=(12, 4))

plt.plot(
    df.loc[inicio:fin].index,
    df.loc[inicio:fin, "LT411"],
    linewidth=0.8,
    label="1s"
)

plt.plot(
    lt411_10s.loc[inicio:fin].index,
    lt411_10s.loc[inicio:fin],
    linewidth=1.3,
    label="10s"
)

plt.plot(
    lt411_1min.loc[inicio:fin].index,
    lt411_1min.loc[inicio:fin],
    linewidth=1.6,
    label="1min"
)

plt.title("LT-411: 1s vs 10s vs 1min — primera hora")
plt.xlabel("Tiempo")
plt.ylabel("Nivel (%)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
inicio = df.index.min()
fin = inicio + pd.Timedelta(minutes=15)

plt.figure(figsize=(12, 4))

plt.plot(
    df.loc[inicio:fin].index,
    df.loc[inicio:fin, "LT411"],
    linewidth=0.8,
    label="1s"
)

plt.plot(
    lt411_10s.loc[inicio:fin].index,
    lt411_10s.loc[inicio:fin],
    linewidth=1.3,
    label="10s"
)

plt.plot(
    lt411_30s.loc[inicio:fin].index,
    lt411_30s.loc[inicio:fin],
    linewidth=1.5,
    label="30s"
)

plt.title("LT-411: 1s vs 10s vs 30s — primeros 15 minutos")
plt.xlabel("Tiempo")
plt.ylabel("Nivel (%)")
plt.legend()
plt.tight_layout()
plt.show()
```
